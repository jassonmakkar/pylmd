```python
import pooch
import time
from pylmd import lmd
```

```python
max_time = 2**3
```

```python
EXAMPLE_DATA = pooch.create(
    path=pooch.os_cache("scverse_tutorials"),
    base_url="doi:10.6084/m9.figshare.22716739.v1/",
)
EXAMPLE_DATA.load_registry_from_doi()

path = EXAMPLE_DATA.fetch('s1d1_filtered_feature_bc_matrix.h5')
```

```python
start_time = time.time()
LMDs = lmd.pyLMD(path, max_time = max_time, device = 'cpu')
end_time = time.time()
execution_time = end_time - start_time
print(f"{max_time} Diffusion Execution time: {execution_time} seconds")
```

```python
start_time = time.time()
LMDs = lmd.pyLMD(path, max_time = max_time, device = 'gpu')
end_time = time.time()
execution_time = end_time - start_time
print(f"{max_time} Diffusion Execution time: {execution_time} seconds")
```
